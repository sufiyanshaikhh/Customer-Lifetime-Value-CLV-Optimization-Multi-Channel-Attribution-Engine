Sample data record to test the script under uncertainity.

```python
test_df = customers.copy()

# Introducing a missing value
test_df.loc[0, "customer_city"] = None

# Introducing a duplicate row
test_df = pd.concat(
    [test_df, test_df.iloc[[0]]],
    ignore_index=True
)

# Introducing a blank string
test_df.loc[1, "customer_state"] = "   "
```

```python
inspector = DataQualityInspector(
    test_df,
    primary_keys=["customer_id"]
)

results = inspector.inspect()
```
